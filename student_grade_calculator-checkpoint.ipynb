#Student Grade Calculator

A beginner-friendly Python project that assigns a grade to each student based on their marks using if, elif, and else statements. m

```python
students = ["Asha", "Ravi", "Meena", "Kiran", "Swetha", "Harsh", "Rupesh", "Akhil", "Divya", "Teja"]
marks = [78, 85, 62, 90, 71, 89, 70, 65, 45, 105]
```

```python
def get_grade(mark):
    if mark < 0 or mark > 100:
        return "Invalid"
    elif mark >= 90:
        return "A"
    elif mark >= 80:
        return "B"
    elif mark >= 70:
        return "C"
    elif mark >= 60:
        return "D"
    else:
        return "F"
```

```python
print(f"{'Student':<10}{'Marks':<8}{'Grade'}")
for student, mark in zip(students, marks):
    print(f"{student:<10}{mark:<8}{get_grade(mark)}")
```
